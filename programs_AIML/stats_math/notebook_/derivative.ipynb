```python
def derivative_f(function_name, x, step_size):
  return ((function_name)(x + step_size) - (function_name)(x)) /  ((x + step_size) -(x)) 

def f_x(x):
  return x**2
slope_at_x3=derivative_f(f_x,3,0.001)
print(slope_at_x3)
```
