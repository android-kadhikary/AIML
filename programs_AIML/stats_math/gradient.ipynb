```python
from sklearn.datasets import make_regression
import numpy as np
import matplotlib.pyplot as plt
```

```python
X,y = make_regression(n_samples=100 ,n_features=1, n_targets=1 , n_informative=1, noise=80, random_state= 13)
plt.scatter(X,y)
```

```python
#Ordinary List Square
from sklearn.linear_model import LinearRegression
reg = LinearRegression()
reg.fit(X,y)
#LinearRegression(copy_X=True, fit_intercept=True, n_jobs=None, normalize=False)
print(reg.coef_)
print(reg.intercept_)
```

```python
plt.scatter(X,y)
plt.plot(X,reg.predict(X), color='red')
```

```python
m=78.35
b=100 #setting to 100
```

```python
y_pred = ((m*X)+b).reshape(4)
plt.scatter(X,y)
plt.plot(X,reg.predict(X), color='red', label='OLS')
plt.plot(X,y_pred, color='green', label='b=0')
```

```python
loss_slope=-2* np.sum(y-m*X.ravel()-b)
print ("loss_slope ", loss_slope)
```

```python
#learning rate = 0.1
#Step size = learning rate * loss_slope
learning_rate=0.1
step_size=learning_rate * loss_slope
print ("step_size : ",step_size)
```

```python
b=b-step_size
print ("  b is = ",b)
y_pred1 = ((m*X)+b).reshape(4)
plt.scatter(X,y)
plt.plot(X,reg.predict(X), color='red', label='OLS')
plt.plot(X,y_pred, color='green', label='b=0')
plt.plot(X,y_pred1, color='blue', label='b={}'.format(b))
```
