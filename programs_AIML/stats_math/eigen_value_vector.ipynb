```python
import numpy as np
 # Define the matrix A
A = np.array([[2, 1],
              [1, 2]])
# Compute the eigenvalues using numpy's linalg.eig
eigenvalues, eigenvectors = np.linalg.eig(A)
# Output the eigenvalues and eigenvectors
print("Eigenvalues of A: ",eigenvalues)
print("eigenvectors of A : ",eigenvectors)
```
